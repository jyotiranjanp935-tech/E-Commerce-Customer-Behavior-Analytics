## ORDERS datetime conversion

```python
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp'], errors='coerce'
)

orders['order_approved_at'] = pd.to_datetime(
    orders['order_approved_at'], errors='coerce'
)

orders['order_delivered_carrier_date'] = pd.to_datetime(
    orders['order_delivered_carrier_date'], errors='coerce'
)

orders['order_delivered_customer_date'] = pd.to_datetime(
    orders['order_delivered_customer_date'], errors='coerce'
)

orders['order_estimated_delivery_date'] = pd.to_datetime(
    orders['order_estimated_delivery_date'], errors='coerce'
)
```

## ORDER_ITEMS datetime conversion

```python
order_items['shipping_limit_date'] = pd.to_datetime(
    order_items['shipping_limit_date'], errors='coerce'
)
```

## REVIEWS datetime conversion¶

```python
reviews['review_creation_date'] = pd.to_datetime(
    reviews['review_creation_date'], errors='coerce'
)

reviews['review_answer_timestamp'] = pd.to_datetime(
    reviews['review_answer_timestamp'], errors='coerce'
)
```

```python
orders.info()
```

## Extrcat Date From the orders

```python
orders['order_purchase_date'] = orders['order_purchase_timestamp'].dt.date
orders['order_purchase_month'] = orders['order_purchase_timestamp'].dt.to_period('M')
orders['order_purchase_hour'] = orders['order_purchase_timestamp'].dt.hour
orders['order_purchase_weekday'] = orders['order_purchase_timestamp'].dt.day_name()
```

```python
orders.head()
```

# Delivery delay

```python
orders['delivery_delay_days'] = (
    orders['order_delivered_customer_date']
    - orders['order_estimated_delivery_date']
).dt.days
```

```python
orders.head()
```

# Order Delivery late Flaging

```python
orders["Delivery_status"] = np.where(orders["delivery_delay_days"] > 0, "Late", "On time")
```

```python
orders.sample(n=10)
```

```python
orders["Delivery_status"].value_counts()
```

```python
plt.pie(orders["Delivery_status"].value_counts(),labels=orders["Delivery_status"].value_counts().index,autopct='%1.1f%%')
```

```python
status_counts = orders["Delivery_status"].value_counts()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0.05, 0],          # highlight late deliveries
    shadow=True,
    wedgeprops={'edgecolor': 'black'}
)

plt.title("Order Delivery Status (On-time vs Late)", fontsize=14)
plt.axis('equal')   # perfect circle
plt.show()
```

# Review Score vs Delivery Delay

```python
orders_reviews["delay_bucket"]=pd.cut(orders_reviews["delivery_delay_days"],
    bins=[-100, 0, 5, 15, 100],
    labels=["Early/On-time", "0–5 late", "5–15 late", "15+ late"])
```

```python
orders_reviews.sample(n=6)
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=orders_reviews,
    x="delay_bucket",
    y="review_score"
)

plt.title("Review Score by Delivery Delay Bucket", fontsize=14, weight="bold")
plt.xlabel("Delivery Delay Bucket")
plt.ylabel("Review Score")
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

# Customer Repeat Purchase Rate

```python
customer_orders = (
    orders_customers
    .groupby("customer_unique_id")["order_id"]
    .nunique()
)
customer_orders.head()
```

```python
repeat_pct = (customer_orders > 1).mean() * 100
repeat_pct = round(repeat_pct, 3)
float(repeat_pct)
```

```python
print(f"Repeat Purchase Rate: {repeat_pct:.1f}%")
```

```python
one_time_pct = 100 - repeat_pct
float(one_time_pct)
```

```python
labels = ["Repeat Customers", "One-time Customers"]
values = [repeat_pct, one_time_pct]

plt.figure(figsize=(5,3))
bars=plt.bar(labels, values,color=["#1f77b4", "#ff7f0e"])

plt.title("Customer Repeat Purchase Rate", fontsize=14, weight="bold")
plt.ylabel("Percentage of Customers (%)")
plt.ylim(0, 100)
legend_labels = [
    f"Repeat Customers: {repeat_pct:.1f}%",
    f"One-time Customers: {one_time_pct:.1f}%"
]
plt.legend(bars, legend_labels, loc="upper left")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend
plt.tight_layout()
plt.show()
```
